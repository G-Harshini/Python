```python
#sum of digits in a number

n=int(input("Enter a number"))
total=0
while(n>0):
    digit=n%10
    total=total+digit
    n=n//10
print(total)
```

```python
#Reverse the number
n=int(input("ener number"))
reverse=0
while n>0:
    digit=n%10
    n=n//10
    reverse=reverse*10+digit
print(reverse)
```

```python
n=int(input("ener number"))
original= n
reverse=0
while n>0:
    digit=n%10
    reverse=reverse*10+digit
    n=n//10
if original==reverse:
    print("palindrome")
else:
    print("not palindrome")
```
